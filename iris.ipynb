## O Danych

Zbiór danych zawiera informacje o trzech gatunkach irysów: Iris setosa, Iris versicolor, i Iris virginica.

Dane obejmują pomiary czterech cech: długość i szerokość działki kielicha oraz długość i szerokość płatka.

Każdy wiersz w zbiorze danych reprezentuje pojedynczy kwiat, a wartości pomiarów są podane w centymetrach.

Zbiór składa się z 150 próbek, po 50 dla każdego gatunku, i jest szeroko stosowany jako podstawowy zbiór danych do
testowania algorytmów klasyfikacji oraz w nauce o danych i uczeniu maszynowym.

Kolumny:
* **długość kielicha (sepal length)** - Długość kielicha w cm
* **szerokość kielicha (sepal width)** - Szerokość kielicha w cm
* **długość płatka (petal length)** - Długość płatka w cm
* **szerokość płatka (petal width)** - Szerokość płatka w cm
* **klasa (class)** - Klasa irysa (setosa, versicolor, virginica)

# Wizualizacja danych

```python
#  by ładniej zwizualizowąc dane na wykresach zmieniam nazwy poszczegolnych kolumn na bardziej przyjazne:
df = df.rename(columns = {'długość kielicha (sepal length)': 'długość_kielicha', 'szerokość kielicha (sepal width)': 'szerokość_kielicha',
                          'długość płatka (petal length)': 'długość_płatka', 'szerokość płatka (petal width)': 'szerokość_płatka', 
                          'klasa (class)': 'gatunek'})
```

```python
df.plot(kind='box')
```

```python
df['długość_płatka'].hist(legend=True, grid=False)
```

```python
df['szerokość_płatka'].hist(legend=True, grid=False)
```

```python
df['długość_kielicha'].hist(legend=True, grid=False)
```

```python
df['szerokość_kielicha'].hist(legend=True, grid=False)
```

```python
# Na tym etapie zorwientowałem się, ze histogram w pandasie ma za mało opcji wyświetlania, stąd zapytanie do chataGPT jak najlepiej zobracować dane w Pythonie i została zaproponowana biblioteka matplotlib 
#średnia długość płatka dla każdego gatunku
df.groupby("gatunek")["długość_płatka"].mean().plot(kind="bar")

plt.xlabel("Gatunek")
plt.ylabel("Średnia długość płatka")
plt.title("Średnia długość płatka wg gatunku")
plt.show()
```

```python
df.groupby("gatunek")["szerokość_płatka"].mean().plot(kind="bar")

plt.xlabel("Gatunek")
plt.ylabel("Średnia szerokość płatka")
plt.title("Średnia szerokość płatka wg gatunku")
plt.show()
```

```python
df.groupby("gatunek")["długość_kielicha"].mean().plot(kind="bar")

plt.xlabel("Gatunek")
plt.ylabel("Średnia długość kielicha w centymetrach")
plt.title("Średnia długość kielicha wg gatunku")
plt.show()
```

```python
df.groupby("gatunek")["szerokość_kielicha"].mean().plot(kind="bar")

plt.xlabel("Gatunek")
plt.ylabel("Średnia szerokość kielicha w centymetrach")
plt.title("Średnia szerokość kielicha wg gatunku")
plt.show()
```

   ### Długość płatka

Histogram przedstawia rozkład długości płatków w badanej próbie (150 rekordów). Widać wyraźnie dwie grupy wartości:
pierwsza grupa z płatkami krótkimi, w zakresie 1–2 cm - najwięcej obserwacji (ponad 35 wystąpień),
druga grupa z płatkami dłuższymi: od 3 cm do 7 cm, rozłożone bardziej równomiernie, z największą częstością w przedziale 4–5 cm.
Rozkład sugeruje, że w zbiorze znajdują się różne gatunki roślin – jeden gatunek ma wyraźnie krótsze płatki, a pozostałe dwa charakteryzują się dłuższymi.
Wykres słupkowy dla średniej długości płatka sugeruje, że do najliczniejszej, pierwszej grupy należy gatunek *iris-sectosa*. 


### Szerokość płatka

Podobnie jak z długością płatków, tak z szerokością również występują dwie grupy. Pierwsza o szerokości płatka od 0.1 cm do 0.4 cm oraz druga o szerokościach płatków 1.2 cm do 2.1 cm
Również, jak z wartością długości płatków irysów, tak i w tym wypadku, na wykresie słupkowym, przedstawiającym średnią szerokość płatków najmniejszą szerokością płatka cechują się *iris-sectosa*, co jest ponownie cechą szczególną umożliwiająca ich szybkie rozpoznanie. 


### Długość kielicha

Histogram zbliżony do rozkładu normalnego. Najwięcej wartości znajduje się w przedziale od 4.6 cm do 5.7 cm.

Rozkład nie jest idealnie symetryczny - istnieją wartości powyżej 7 cm, ale są rzadkie. 


### Szerokość kielicha

Histogram również zbliżony do rozkładu normalnego, jednakże na podstawie wykresu pudełkowego widać, że jest kilka wartości odstających (powyżej 4cm)

Najwięcej wartości znajdziemy w przedziale 2.5 do 3.2 cm. 

Te kwiaty, posiadające najszersze kielichy mogą być zaliczone do jednego gatunku i znów będzie to *iris-sectosa (na podstawie wykresu średniej szerokości kielicha).





```python
df['gatunek'].value_counts()
```

```python
df['gatunek'].hist(grid=False)
```

```python
#zaokrąglam wartości do jednej miejsca po przecinku, gdyż wartości są na tyle różne dla różnych gatunków, że takie duże zaokrąglenie wystarczy by wysnuć wnioski
najwieksze_wartosci = df.groupby('gatunek')[['długość_płatka', 'szerokość_płatka', 'długość_kielicha', 'szerokość_kielicha']].max().round(1)
najwieksze_wartosci
#odkomentuj poniżej by zobaczyc wykres słupkowy dla maksymalnych wartości długość płatka, szerokość płatka, długość kielicha, szerokość kielicha
#najwieksze_wartosci.plot(kind='bar', ylabel='cm')
```

```python
#zaokrąglam wartości do jednej miejsca po przecinku, gdyż wartości są na tyle różne dla różnych gatunków, że takie duże zaokrąglenie wystarczy by wysnuć wnioski
najmniejsze_wartosci = df.groupby('gatunek')[['długość_płatka', 'szerokość_płatka', 'długość_kielicha', 'szerokość_kielicha']].min().round(1)
najmniejsze_wartosci
#odkomentuj poniżej by zobaczyc wykres słupkowy dla maksymalnych wartości długość płatka, szerokość płatka, długość kielicha, szerokość kielicha
#najmniejsze_wartosci.plot(kind='bar', ylabel='cm')
```

```python
#zaokrąglam wartości do jednej miejsca po przecinku, gdyż wartości są na tyle różne dla różnych gatunków, że takie duże zaokrąglenie wystarczy by wysnuć wnioski
srednie_wartosci = df.groupby('gatunek')[['długość_płatka', 'szerokość_płatka', 'długość_kielicha', 'szerokość_kielicha']].mean().round(1)
srednie_wartosci
#odkomentuj poniżej by zobaczyc wykres słupkowy dla srednich  wartości długość płatka, szerokość płatka, długość kielicha, szerokość kielicha
#srednie_wartosci.plot(kind='bar', ylabel='cm')
```

```python
mediana_wartosci = df.groupby('gatunek')[['długość_płatka', 'szerokość_płatka', 'długość_kielicha', 'szerokość_kielicha']].median().round(1)
mediana_wartosci
#odkomentuj poniżej by zobaczyc wykres słupkowy dla mediany  wartości długość płatka, szerokość płatka, długość kielicha, szerokość kielicha
#mediana_wartosci.plot(kind='bar', ylabel='cm')
```

```python
srednie_wartosci = df.groupby('gatunek')[['długość_płatka', 'szerokość_płatka', 'długość_kielicha', 'szerokość_kielicha']].mean()
srednie_wartosci.plot(kind='bar', ylabel='cm')
```
